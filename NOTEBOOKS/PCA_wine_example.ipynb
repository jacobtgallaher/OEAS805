## Principal Component Analysis

Principal Component Analysis (PCA) is a statistical technique used to reduce the dimensionality of high-dimensional data while retaining as much information as possible. The goal of PCA is to find a new set of variables, called principal components, that captures most of the variance in the original data.

In other words, PCA takes a dataset with many variables and transforms it into a smaller set of variables that still contains the most important information about the data. This is accomplished by finding linear combinations of the original variables that explain the most variation in the data. These linear combinations are the principal components.

The first principal component captures the most variation in the data, and each subsequent component captures as much variation as possible while being orthogonal (i.e., uncorrelated) to the previous components. The number of principal components to keep is typically determined by the amount of variance they explain, with a common rule of thumb being to keep enough components to explain at least 70% to 80% of the variance.


```python
# import libraries/packages
%pip install plotly
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import plotly.express as px
```

```python
# load in the wine data
wine = load_wine()

data = pd.DataFrame(wine.data, columns = wine.feature_names)
data
```

```python
scaler = StandardScaler()
scaler.fit(data)
data_scaled = scaler.transform(data)

pca = PCA(n_components = 0.95) # we tell it the target percentage of cumulative variance to preserve.

pca.fit(data_scaled)
components = pca.fit_transform(data_scaled)

print(pca)
```

```python
fig, ax = plt.subplots()
# Create an index array corresponding to each principal component (1 through N).
xi = range(1, pca.n_components_+1) 
y = np.cumsum(pca.explained_variance_ratio_)

# The Principal Component Index (x-axis) represents the rank order of the newly created orthogonal axes (components) in Principal Component Analysis (PCA).
# In this  specific dataset, PCA identified a given number of principal components.

# Individual variance
plt.bar(xi, pca.explained_variance_ratio_, alpha=0.5, align='center',label='Individual explained variance')

# Cumulative variance
plt.plot(xi, y , '-o', label='Cumulative explained variance',color='red')

# 80% Variance threshold
plt.axhline(0.8, linestyle = '--') # Draws a dashed horizontal threshold line at 0.8 (80% explained variance) to evaluate feature reduction targets

plt.ylabel('Explained variance ratio')
plt.xlabel('Principal component index')
plt.show()
plt.show()
```

```python

loadings = pca.components_.T * np.sqrt(pca.explained_variance_)

features = wine.feature_names

fig = px.scatter(components, x=0, y=1, color = wine.target) # try different values of x and y (1 to 10)

for i, feature in enumerate(features):
    fig.add_annotation(
        ax=0, ay=0,
        axref="x", ayref="y",
        x=loadings[i, 0],
        y=loadings[i, 1],
        showarrow=True,
        arrowsize=2,
        arrowhead=2,
        xanchor="right",
        yanchor="top"
    )
    fig.add_annotation(
        x=loadings[i, 0],
        y=loadings[i, 1],
        ax=0, ay=0,
        xanchor="center",
        yanchor="bottom",
        text=feature,
        yshift=5,
    )
fig.show()
```
